# String for punctuation remove

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import string
```

# As it was a txt file so we must mention how values are seperated inside file
# names=(Before and after(;) headings of entries)

```python
df=pd.read_csv('train.txt',sep=';',header=None,names=['text','emotion'])
```

```python
df.head(5)
```

```python
df.isnull().sum()
```

```python
unqEmo=df['emotion'].unique()
```

# Dictionary creation for all unique values in output(One-Hot  encoding)

```python
Emotions_numbers={}
i=0
for emo in unqEmo:
  Emotions_numbers[emo]=i
  i+=1

```

# Mapping means creating relation with dictionary and dataset

```python
 df['emotion']=df['emotion'].map(Emotions_numbers)
```

```python
df
```

# Converting text to lowercase
# Lambda: to create a small function

```python
df['text']=df['text'].apply(lambda x:x.lower())
```

# Removing Punctuation

```python
def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punc)
```

# Removing Numbers

```python
def remove_numbers(txt):
  new=""
  for i in txt:
    if not i.isdigit():
      new=new+i
  return new
```

```python
df['text']=df['text'].apply(remove_numbers)
```

# URLS AND LINKS REMOVING(This dataset is already cleaned because it's in kaggle) Just wrote this as a practice

```python
import re

def remove_urls(text):
    url_pattern = re.compile(r'https?://\S+|www\.\S+')
    return url_pattern.sub(r'', text)
```

```python
df['text'] = df['text'].apply(remove_urls)
```

# Removing emojis and HTML Tags
# Emojis have UNICODE not ASCII value

```python
def remove_emojis(txt):
  new=""
  for i in txt:
    if i.isascii():
         new+=i
  return new
```

```python
df['text']=df['text'].apply(remove_emojis)
```

# Stopwords Removing:Repeating words like(is,was,the) which generally give much meaning to sentence

# NLTK: Library good for stopwords
# corpus: For big collection of text
# Token:collection of it makes text(Individual symbol)
# Tokenization: process of converting text to tokens(individual values)

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

# punkt: does tokenization

```python
nltk.download('stopwords')
nltk.download('punkt_tab')
```

# All stopwords

```python
stop_words=set(stopwords.words('english'))
stop_words
```

```python
len(stop_words)
```

# First go to every text step by step and extract text only
# Form an array of text in sentence seperatin each text with ""
# Then find stopword if there
# remove it

```python
def remove(txt):
  words=word_tokenize(txt)
  cleaned=[]
  for i in words:
    if i not in stop_words:
      cleaned.append(i)
  return ' '.join(cleaned)




```

```python
df['text']=df['text'].apply(remove)
```

# Model Creation
# Data Split
# Don't confuse where is X and Y because always Input=x,Output=y

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)

```

# Vectorization
# Input or text part(X_Train,X_Test)
# Output or Emotion Part   (Y_Train,Y_Test)
# countvectorizer=BagOW

```python
from sklearn.feature_extraction.text import TfidfVectorizer,CountVectorizer
vect=CountVectorizer()
```

# Using Bag Of Words model creation
# Naives Bayes

# Always fit (X_Train,Test) in Train Test split

```python
Xtrain_vect=vect.fit_transform(X_train)
Xtest_vect=vect.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
model=MultinomialNB()
```

```python
model.fit(Xtrain_vect,y_train)
```

```python
y_pred=model.predict(Xtest_vect)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,y_pred))
```
