# BagOfWords

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents=[
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]
```

# Normal One

```python
# vectorizer=CountVectorizer()
```

# Using n_gram

```python
vectorizer=CountVectorizer(ngram_range=(2,2))
```

```python
X=vectorizer.fit_transform(documents)
```

```python
print("vocabularu:",vectorizer.get_feature_names_out())
print("\nBoW Matrix:\n",X.toarray())
```

# TFIDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
documents1=[
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]
```

```python
vectorizer=TfidfVectorizer()
```

```python
X1=vectorizer.fit_transform(documents1)
```

```python
print("vocabularu:",vectorizer.get_feature_names_out())
print("\nBoW Matrix:\n",X1.toarray())
```
